```python
import cv2
import numpy as np
from google.colab import drive
from google.colab.patches import cv2_imshow

# Mount Google Drive
drive.mount('/content/drive')

# Read image
img = cv2.imread("/content/drive/MyDrive/opencv/umbrella.jpg")

# Check if image loaded
if img is None:
    print("Image not found!")
else:
    # Convert to grayscale
    img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply CLAHE
    clahe = cv2.createCLAHE(
        clipLimit=2.0,
        tileGridSize=(8, 8)
    )

    result = clahe.apply(img_grey)

    # Display original
    print("Original Image:")
    cv2_imshow(img)

    # Display grayscale
    print("Grayscale Image:")
    cv2_imshow(img_grey)

    # Display CLAHE result
    print("CLAHE Image:")
    cv2_imshow(result)
```

```python
#edges 
edges1 = cv2.Canny(result, threshold1=100, threshold2=200)
edges2 = cv2.Canny(img, threshold1=50, threshold2=150)
edges3 = cv2.Canny(img_grey, threshold1=200, threshold2=300)


# Display edges
cv2_imshow(edges1)
cv2_imshow(edges2)
cv2_imshow(edges3)
```
