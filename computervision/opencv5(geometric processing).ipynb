```python
import cv2
import numpy as np 

```

```python
from google.colab import drive
drive.mount('/content/drive')

img=cv2.imread("/content/drive/MyDrive/opencv/umbrella.jpg")
```

```python
img.shape[1], img.shape[0]
```

```python
#TRANSLATION

tx, ty = 50, 30  # shift 50px right, 30px down
M = np.float32([[1, 0, tx], [0, 1, ty]])
shifted = cv2.warpAffine(img, M, (540, 360))
from google.colab.patches  import  cv2_imshow
cv2_imshow(shifted)
```

```python
#ROTATION

center = (img.shape[1]//2, img.shape[0]//2)
M = cv2.getRotationMatrix2D(center, angle=45, scale=1.0)
rotated = cv2.warpAffine(img, M, (img.shape[1], img.shape[0]))
cv2_imshow(rotated)
```

```python
#SCALING 
resized = cv2.resize(img, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)
cv2_imshow(resized)
```
