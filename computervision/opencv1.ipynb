```python
img=cv2.imread("/content/drive/MyDrive/opencv/chobooo.png")
print(img)
```

```python
print(img.shape)   # (height, width, 3)
print(img.dtype)   # uint8 -> values 0-255
print(type(img))   # <class 'numpy.ndarray'>

# Indexing is (row, col) = (y, x) — NOT (x, y)
pixel = img[100, 200]   # pixel at row=100, col=200 -> returns [B, G, R]

# Cropping is just slicing: img[y1:y2, x1:x2]
cropped = img[50:200, 100:300]
```

```python
from google.colab.patches import cv2_imshow

cv2_imshow(img)
```

```python
#this is BGR image, so the channels are in reverse order. 
# to test we will display it in matplotlib which uses RGB format.
import matplotlib
matplotlib.pyplot.imshow(img)
```

```python

#we need to convert it to RGB format before displaying it in matplotlib.
#Important parameter: cv2.cvtColor(src, code) — code is one of ~200 conversion 
#constants (COLOR_BGR2RGB, COLOR_BGR2GRAY, COLOR_BGR2HSV, etc.). 
#You'll use this function constantly; only a handful of codes matter in practice.

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
matplotlib.pyplot.imshow(img_rgb)
```

```python
#converting from uint8 to float32 and normalizing the values to [0, 1] range.
img_float = img.astype('float32') / 255.0
print(img_float.dtype)              # float32
print(img_float.min(), img_float.max())   # ~0.0, ~1.0
```

```python
#conversion to greyscale 
img_greyscale = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(img_greyscale)
```

```python
#what does imwrite do ??
# cv2.imwrite(filename, image) - saves an image to a file
cv2.imwrite('output.png', img)
cv2.imwrite('output.jpg', img, [cv2.IMWRITE_JPEG_QUALITY, 90])
```
